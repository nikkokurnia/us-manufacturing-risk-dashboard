```python
new_orders_path = "../data_raw/historical_new_orders.xlsx"
shipments_path = "../data_raw/historical_shipments.xlsx"
unfilled_orders_path = "../data_raw/historical_unfilled_orders.xlsx"
total_inventories_path = "../data_raw/historical_total_inventories.xlsx"
inventories_to_shipments_path = "../data_raw/historical_inventories_to_shipments.xlsx"
unfilled_orders_to_shipments_path = "../data_raw/historical_unfilled_orders_to_shipments.xlsx"
```

```python
historical_paths = [
    new_orders_path,
    shipments_path,
    unfilled_orders_path,
    total_inventories_path,
    inventories_to_shipments_path,
    unfilled_orders_to_shipments_path
]

for file_path in historical_paths:
    excel_file = pd.ExcelFile(file_path)
    print(file_path, excel_file.sheet_names)
```

```python
def clean_historical_m3_file(file_path, value_name):
    raw = pd.read_excel(
        file_path,
        sheet_name=0,
        header=None
    )

    raw.columns = [
        "series_code",
        "year",
        "jan",
        "feb",
        "mar",
        "apr",
        "may",
        "jun",
        "jul",
        "aug",
        "sep",
        "oct",
        "nov",
        "dec"
    ]

    clean = raw.copy()

    clean["year"] = pd.to_numeric(clean["year"], errors="coerce")
    clean = clean.dropna(subset=["year"])
    clean["year"] = clean["year"].astype(int)

    long = clean.melt(
        id_vars=["series_code", "year"],
        value_vars=[
            "jan", "feb", "mar", "apr", "may", "jun",
            "jul", "aug", "sep", "oct", "nov", "dec"
        ],
        var_name="month",
        value_name=value_name
    )

    month_map = {
        "jan": 1,
        "feb": 2,
        "mar": 3,
        "apr": 4,
        "may": 5,
        "jun": 6,
        "jul": 7,
        "aug": 8,
        "sep": 9,
        "oct": 10,
        "nov": 11,
        "dec": 12
    }

    long["month_num"] = long["month"].map(month_map)

    long["date"] = pd.to_datetime(
        dict(
            year=long["year"],
            month=long["month_num"],
            day=1
        )
    )

    long[value_name] = pd.to_numeric(long[value_name], errors="coerce")

    long = long[
        ["series_code", "date", "year", "month_num", value_name]
    ]

    return long
```

```python
new_orders_clean = clean_historical_m3_file(
    new_orders_path,
    "new_orders"
)

shipments_clean = clean_historical_m3_file(
    shipments_path,
    "shipments"
)

unfilled_orders_clean = clean_historical_m3_file(
    unfilled_orders_path,
    "unfilled_orders"
)

total_inventories_clean = clean_historical_m3_file(
    total_inventories_path,
    "total_inventories"
)

inventories_to_shipments_clean = clean_historical_m3_file(
    inventories_to_shipments_path,
    "inventories_to_shipments"
)

unfilled_orders_to_shipments_clean = clean_historical_m3_file(
    unfilled_orders_to_shipments_path,
    "unfilled_orders_to_shipments"
)

display(new_orders_clean.head())
display(shipments_clean.head())
display(unfilled_orders_clean.head())

print("New Orders:", new_orders_clean.shape)
print("Shipments:", shipments_clean.shape)
print("Unfilled Orders:", unfilled_orders_clean.shape)
print("Total Inventories:", total_inventories_clean.shape)
print("Inventories to Shipments:", inventories_to_shipments_clean.shape)
print("Unfilled Orders to Shipments:", unfilled_orders_to_shipments_clean.shape)
```

```python
new_orders_clean.to_csv("../data_cleaned/new_orders_cleaned.csv", index=False)
shipments_clean.to_csv("../data_cleaned/shipments_cleaned.csv", index=False)
unfilled_orders_clean.to_csv("../data_cleaned/unfilled_orders_cleaned.csv", index=False)
total_inventories_clean.to_csv("../data_cleaned/total_inventories_cleaned.csv", index=False)
inventories_to_shipments_clean.to_csv("../data_cleaned/inventories_to_shipments_cleaned.csv", index=False)
unfilled_orders_to_shipments_clean.to_csv("../data_cleaned/unfilled_orders_to_shipments_cleaned.csv", index=False)

print("All cleaned CSV files saved.")
```

```python
# Load cleaned files
new_orders = pd.read_csv("../data_cleaned/new_orders_cleaned.csv")
shipments = pd.read_csv("../data_cleaned/shipments_cleaned.csv")
unfilled_orders = pd.read_csv("../data_cleaned/unfilled_orders_cleaned.csv")
total_inventories = pd.read_csv("../data_cleaned/total_inventories_cleaned.csv")
inventories_to_shipments = pd.read_csv("../data_cleaned/inventories_to_shipments_cleaned.csv")
unfilled_orders_to_shipments = pd.read_csv("../data_cleaned/unfilled_orders_to_shipments_cleaned.csv")

# Merge all datasets together
master = new_orders.merge(
    shipments,
    on=["series_code", "date", "year", "month_num"],
    how="left"
)

master = master.merge(
    unfilled_orders,
    on=["series_code", "date", "year", "month_num"],
    how="left"
)

master = master.merge(
    total_inventories,
    on=["series_code", "date", "year", "month_num"],
    how="left"
)

master = master.merge(
    inventories_to_shipments,
    on=["series_code", "date", "year", "month_num"],
    how="left"
)

master = master.merge(
    unfilled_orders_to_shipments,
    on=["series_code", "date", "year", "month_num"],
    how="left"
)

# Convert date column back to datetime
master["date"] = pd.to_datetime(master["date"])

display(master.head())
display(master.tail())
print(master.shape)
print(master.columns)
```

```python
master.to_csv("../data_cleaned/m3_master_supply_chain_dataset.csv", index=False)

print("Master dataset saved.")
```
